```python
#Factorial
print("Enter the number: ")
number = int(input())

Factorial = 1
i = 1
while i <= number:
  Factorial = Factorial * i
  i = i + 1

print("Factorial: ",Factorial)
```

```python
#Factorial
number = int(input("Enter the number: "))

fact = 1
if number < 0:
  print("Not defined")
else:
  while number > 0:
    fact = fact * number
    number = number - 1
  print("Factorial: ",fact)
```

```python
#Reverse the digit in the given number
#Using the while loop

num = int(input('Enter the number: '))
absNum = abs(num)
if num >= 0:
  rev = num % 10
  num = num // 10
  while num > 0:
    r = num % 10
    num = num // 10
    rev = rev * 10 + r
  print(rev)
else:
  rev = absNum % 10
  absNum = absNum // 10
  while absNum > 0:
    r = absNum % 10
    absNum = absNum // 10
    rev = rev * 10 + r
  print(rev - 2*rev)
```

```python
#Find whether the entered numbe is palindrome or not

num = int(input('Enter the number: '))
absNum = abs(num)
rev = absNum % 10
absNum = absNum // 10
while absNum > 0:
    r = absNum % 10
    absNum = absNum // 10
    rev = rev * 10 + r
if num < 0:
  rev = rev - 2 * rev
print(rev)
if num == rev:
  print('Palindrom')
else:

  print('Not a palindrom')
```
