```python
#Factorial
print("Enter the number: ")
number = int(input())

Factorial = 1
i = 1
while i <= number:
  Factorial = Factorial * i
  i = i + 1

print("Factorial: ",Factorial)
```

```python
#Factorial
number = int(input("Enter the number: "))

fact = 1
if number < 0:
  print("Not defined")
else:
  while number > 0:
    fact = fact * number
    number = number - 1
  print("Factorial: ",fact)
```

```python
#Reverse the digit in the given number
#Using string method

number = int(input('Enter the number: '))
str_number = str(number)
if number >= 0:
  print(str_number[::-1])
else:
  take_zeroth = str_number[0]
  rev_num = str_number[1:]
  print(take_zeroth+''+rev_num[::-1])
```

```python
#Find whether the entered numbe is palindrome or not

num = int(input('Enter the number: '))
absNum = abs(num)
rev = absNum % 10
absNum = absNum // 10
while absNum > 0:
    r = absNum % 10
    absNum = absNum // 10
    rev = rev * 10 + r
if num < 0:
  rev = rev - 2 * rev
print(rev)
if num == rev:
  print('Palindrom')
else:

  print('Not a palindrom')
```
