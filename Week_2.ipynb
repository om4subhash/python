The algorithm for this problem is given as part of the template code.Implement the algorithm in Python.

```python
start = input()
end = input()
h_position = 'ABCDEFGH'

start_hor = start[0]
end_hor = end[0]
start_var = int(start[1])
end_var = int(end[1])

start_hor_index = h_position.index(start_hor)
end_hor_index = h_position.index(end_hor)

if abs(start_hor_index - end_hor_index) == abs(start_var - end_var):
  print('Yes')
else:
  print('No')
```

Accept a string as input and print the vowels present in the string in alphabetical order. If the string doesn't contain any vowels, then print the string none as output. Each vowel that appears in the input string — irrespective of its case — should appear just once in lower case in the output.

```python
string = input()
vowel = ''

if 'a' in string.lower():
  vowel += 'a'
if 'e' in string.lower():
  vowel += 'e'
if 'i' in string.lower():
  vowel += 'i'
if 'o' in string.lower():
  vowel += 'o'
if 'u' in string.lower():
  vowel += 'u'

if vowel == '':
  print('None')
else:
  print(vowel)

```
